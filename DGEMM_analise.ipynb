# Analise dos resultados do DGEMM

Este notebook utiliza somente os CSVs produzidos para as tres partes desta execucao. Os arquivos antigos do repositorio nao sao carregados. Ele resume o baseline, SIMD, blocking e OpenMP; as linhas marcadas como preliminar devem ser tratadas separadamente.

```python
import csv
import statistics
from pathlib import Path

def load_csv(filename):
    with Path(filename).open(newline='', encoding='utf-8') as file:
        return list(csv.DictReader(file))

parte1 = load_csv('resultados_parte1.csv')
parte2 = load_csv('resultados_parte2.csv')
parte3 = load_csv('resultados_parte3.csv')
print(f'Parte 1: {len(parte1)} registros')
print(f'Parte 2: {len(parte2)} registros')
print(f'Parte 3: {len(parte3)} registros')
```

```python
print('Resultados da Parte 1')
for row in parte1:
    print(f"N={row['n']}: {row['multiplicacoes']} multiplicacoes, {row['gflops']} GFLOPS")

baseline_n32 = next(row for row in parte1 if row['n'] == '32')
print(f"\nBaseline em N=32: {baseline_n32['gflops']} GFLOPS")
```

```python
print('Resultados da Parte 2')
for row in parte2:
    print(f"{row['versao']}: {row['gflops']} GFLOPS, speedup {row['speedup']}x")

gflops_parte2 = [float(row['gflops']) for row in parte2]
print(f"Media dos valores de GFLOPS da Parte 2: {statistics.mean(gflops_parte2):.2f}")
```

```python
try:
    import matplotlib.pyplot as plt
    labels = [row['versao'] for row in parte2]
    values = [float(row['gflops']) for row in parte2]
    plt.figure(figsize=(8, 4))
    plt.bar(labels, values, color=['#6b7280', '#2563eb', '#059669'])
    plt.ylabel('GFLOPS')
    plt.title('Parte 2 - Comparacao de desempenho em N=32')
    plt.xticks(rotation=15, ha='right')
    plt.tight_layout()
    plt.show()
except ImportError:
    print('matplotlib nao esta instalado; os dados continuam disponiveis nas tabelas e no CSV.')
```

```python
print('Resultados da Parte 3')
for row in parte3:
    print(f"{row['versao']} ({row['threads']} threads): {row['gflops']} GFLOPS, speedup vs blocking {row['speedup_vs_blocking']}x")

openmp = [row for row in parte3 if row['versao'] == 'OpenMP']
print('\nGFLOPS do OpenMP por numero de threads:')
for row in openmp:
    print(f"{row['threads']} threads: {row['gflops']} GFLOPS")
```
